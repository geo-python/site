## AI and LLM

AI, or Artificial Intelligence, refers to the field of computer science dedicated to developing systems and algorithms that enable machines to perform tasks typically requiring human intelligence, such as problem-solving, learning, and natural language understanding. On the other hand, LLM, or Large Language Models, are a specific category of AI models designed to process and generate human language text. ChatGPT (Chat Generative Pre-trained Transformer) is one common example of an LLM, created by OpenAI. Such tools excel in understanding and generating human-like text, making it useful for tasks like chatbots, virtual assistants, and content generation. It's just one example, other widely used LLM-based tools include Claude, Gemini, and AI features built directly into code editors like GitHub Copilot, Claude Code, and Cursor.

## Using an AI chat assistant

A prompt is a specific instruction, query, or input provided to a computer program or artificial intelligence system to elicit a desired response or output. In the context of language models like Copilot, a prompt typically consists of a text-based request or question that users provide to the model. The model generates a response based on the content and context of the prompt. Prompts are used to interact with AI systems, instruct them on what task to perform, or seek information or assistance from them. The quality and clarity of the prompt can significantly impact the accuracy and relevance of the AI's response.

Lets start with an example prompt we can type in:

```

```

Response:

````python
def add_numbers(a, b):
    return a + b
````

OK, so we clearly got back a Python function that appears to work to add two numbers. Cool!

Most AI chat assistants remember your conversation (i.e., history of prompts). So you can ask it to modify a previously provided response for you. Let's try this with another prompt.

Response:

````python
def add_numbers(a, b):

    return a + b

num1 = 5
num2 = 3
result = add_numbers(num1, num2)
formatted_string = f"The sum of {num1} and {num2} is {result}."
print(formatted_string)
````

OK, so it remembered our function and did as requested.

Let's try another prompt:

```
Add inline comments to the code
```

Response:

````python
# Define a function that takes two numbers as input and returns their sum.
def add_numbers(a, b):
    """
    This function takes two numbers as input and returns their sum.
    
    :param a: The first number to be added.
    :param b: The second number to be added.
    :return: The sum of the two input numbers.
    """
    return a + b

# Define two numbers to be added.
num1 = 5
num2 = 3

# Call the add_numbers function and store the result in the 'result' variable.
result = add_numbers(num1, num2)

# Using an f-string to format the output and include variable values.
formatted_string = f"The sum of {num1} and {num2} is {result}."

# Print the formatted string.
print(formatted_string)
````

Let's try another prompt:

```

```

Response:

````python
def fahrenheit_to_celsius(fahrenheit):
    """
    Converts a temperature from degrees Fahrenheit to degrees Celsius.

    :param fahrenheit: Temperature in degrees Fahrenheit
    :return: Temperature in degrees Celsius
    """
    celsius = (fahrenheit - 32) * 5/9
    return celsius
````

### Writing good prompts

Remember that you are talking with a machine in your prompts. So it is very important for you to provide prompts that are **clear** and have **clear objectives**. In your prompts, try to provide as much **context** and **instruction** as possible.

Here's an example of a bad prompt:

```
A function to convert Kelvin to Newton
```

Response:

````python
def kelvin_to_newton(mass_kg, acceleration_m_s2):
    force_newton = mass_kg * acceleration_m_s2
    return force_newton
````

This doesn't look correct, as we were aiming to convert from Kelvins to degrees Newton, as we did in the functions lesson.

Let's write this better by adding more context:

```
A function to convert temperature in Kelvin to Newton
```

````python
def kelvin_to_newton(kelvin_temperature):
    newton_temperature = kelvin_temperature * 0.33
    return newton_temperature
````

OK, so now we get a better looking function that at least has something about a temperature conversion.

## A different kind of mistake: hallucinated code

The examples above involved an AI misunderstanding *what* you wanted. There's a second, sneakier category of mistake: the AI confidently using a function, method, or parameter that **doesn't actually exist**. This is called a *hallucination*, and unlike the counting mistakes above, it doesn't reliably get better as models improve -- it can happen with even the most capable current models, especially for less common libraries.

This matters a lot for specialized work such as GIS, because geospatial libraries like `geopandas` and `shapely` are far less represented in AI training data than, say, general Python or web development code. Let's see an example.

Suppose we ask an AI assistant:

```
Write a function that takes a GeoDataFrame of points and a reference point, and returns the GeoDataFrame with a new column giving the distance in kilometers from each point to the reference point.
```

Response:

````python
# This is what an AI assistant might confidently hand back:
def add_distance_km(gdf, reference_point):
    gdf["distance_km"] = gdf.geometry.distance_km(reference_point)
    return gdf
````

Looks tidy. Let's try running it:

There's no such thing as `.distance_km()` in geopandas -- the AI invented a method name that sounds entirely plausible, because `.distance()` *does* exist. This is the hallmark of a hallucination: it's not a typo, it's a fabricated API that reads as completely natural. Nothing about the code *looks* wrong until you actually run it.

And even the real `.distance()` method hides a second, very GIS-specific trap: it returns distances in whatever unit your GeoDataFrame's CRS uses. If your data is in EPSG:4326 (lat/lon, degrees), `.distance()` gives you degrees -- not meters, and definitely not kilometers. A correct version needs to reproject to a projected CRS first:

````python
# A corrected version
def add_distance_km(gdf, reference_point):
    # Reproject to a projected CRS (meters) before measuring distance -- e.g. UTM
    gdf_projected = gdf.to_crs(gdf.estimate_utm_crs())
    ref_projected = reference_point  # assume already reprojected to match
    gdf["distance_km"] = gdf_projected.geometry.distance(ref_projected) / 1000
    return gdf
````

```{note}
You would only catch either of these problems -- the invented method, and the CRS/units trap -- by already knowing enough geopandas to recognize what a real method looks like, and enough GIS fundamentals to know that distance depends on your coordinate reference system. An AI assistant won't reliably flag this for you; it produced the wrong answer just as confidently as it would produce a right one.
```

## Staying the brain

Notice what happened across all the examples above: misunderstood intent (the Kelvin-to-Newton mix-up), a quiet logic bug (-60°C and -1°C landing in the same class), and an outright fabricated method that only failed when we actually ran it (`.distance_km()`). These aren't edge cases -- they're the normal experience of working with AI-generated code, and they don't all go away as models get smarter. Hallucinated APIs in particular are still common with today's best models, especially for less mainstream libraries like the geospatial ones you'll use in this course.

As a beginner, this should make you *more* cautious about using AI to write your code, not less:

- **You can't catch a mistake you don't have the skill to recognize.** An experienced programmer looked at the -60/-1 bug above and spotted it immediately. Right now, you likely couldn't -- which is exactly why relying on AI at this stage is risky, not convenient.
- **Outsourcing the writing outsources the learning.** The struggle of writing code yourself -- getting it wrong, debugging it, understanding why it broke -- is how you actually build programming skill. If AI writes it for you, you skip the part where the learning happens.
- **Always run and test any code, with values you know the answer to**, if you do use AI-generated code for any reason.
- **Read every line and be able to explain it.** If you can't explain what a piece of code does, don't submit it as your own work.
- **Vague prompts get vague or wrong answers**, and a confident-sounding answer is not the same as a correct one -- AI text reads fluently whether it's right or wrong.
- **You are responsible for the code you submit**, full stop.

The bottom line for this course: build the skill first. AI tools built into editors (Copilot, Claude Code, Cursor, etc.) aren't a shortcut around that -- they're worth learning to use well only *after* you can already do the underlying work yourself.
